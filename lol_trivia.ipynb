```python
from bs4 import BeautifulSoup
import requests

url = "https://wiki.leagueoflegends.com/en-us/Caitlyn"
res = requests.get(url)
soup = BeautifulSoup(res.text, "html.parser")
```

```python
print(soup)
```

```python
soup.find(id="Trivia").next_elements
```

```python
trivia = []

for h2 in soup.find_all('h2'):
    data = [h2]
            
    trivia.append(data)

print(trivia)
```

```python
trivia
```

```python
trivia_h2 = soup.find('span', {'id': 'Trivia'}).find_parent('h2')
print(trivia_h2)
```

```python
content = []
for sibling in trivia_h2.find_next_siblings():
    if sibling.name == 'h2':
        break
    content.append(sibling)

# Print or process the extracted content
for element in content:
    print(element)
```

```python
# Extract text from all elements within the 'Trivia' section
extracted_text = []

for element in content:
    if hasattr(element, 'get_text'):
        extracted_text.append(element.get_text())  # Extract and strip leading/trailing spaces
    elif isinstance(element, str):
        # Handle raw strings that may appear in BeautifulSoup results
        extracted_text.append(element.strip())

# Join all extracted text with spaces
clean_text = " ".join(extracted_text)

# Print the final clean text
print(clean_text)
```

```python
import pandas as pd

df = pd.read_csv("Loldle Data.csv")

list_columns = ["Position", "Species", "Region"] 

for col in list_columns:
    if col in df.columns:
        df[col] = df[col].apply(lambda x: str(x).split(',') if pd.notna(x) else [])

print(df)

# # Export to JSON
# df.to_json("your_file.json", orient="records", indent=2)
```
